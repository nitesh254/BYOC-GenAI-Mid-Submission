# DocuMind AI — V1 Baseline Corpus & Chunking Exploration

This notebook inspects the bundled CC BY 4.0 policy PDFs (`data/sample/`) and validates page extraction and chunk statistics using the modular `src/` package.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.config import load_config, resolve_path
from src.ingestion.pdf_loader import load_pdfs_from_directory
from src.ingestion.chunker import chunk_documents

cfg = load_config()
pages = load_pdfs_from_directory(resolve_path(cfg['paths']['sample_pdf_dir']))
chunks = chunk_documents(
    pages,
    chunk_size_words=cfg['ingestion']['chunk_size_words'],
    chunk_overlap_words=cfg['ingestion']['chunk_overlap_words'],
    min_chunk_words=cfg['ingestion']['min_chunk_words'],
)
print(f'Loaded {len(pages)} PDF pages -> {len(chunks)} overlapping chunks')
```

```python
import pandas as pd

df = pd.DataFrame([c.to_dict() for c in chunks])
display(df.groupby('source').agg(pages=('page', 'nunique'), chunks=('chunk_id', 'count'), avg_words=('word_count', 'mean')))
```
